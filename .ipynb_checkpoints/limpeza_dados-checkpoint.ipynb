```python
# 1. Importando a biblioteca
import pandas as pd

# 2. Lendo o arquivo CSV
df = pd.read_csv("dados.csv", sep=",", engine="python")

# 3. Exibindo informações gerais do DataFrame
print("Informações gerais:")
print(df.info())

# 4. Exibindo primeiras e últimas linhas
print("\nPrimeiras 5 linhas:")
print(df.head())
print("\nÚltimas 5 linhas:")
print(df.tail())

# 5. Criando uma cópia do DataFrame
df_copy = df.copy()

# 6. Substituindo valores nulos da coluna Calories por 0
df_copy["Calories"] = pd.to_numeric(df_copy["Calories"], errors="coerce")
df_copy["Calories"].fillna(0, inplace=True)

# 7. Substituindo valores nulos da coluna Date por '1900/01/01'
df_copy["Date"] = df_copy["Date"].replace("nan", pd.NA)
df_copy["Date"].fillna("1900/01/01", inplace=True)

# 8. Tentando converter Date para datetime (dará erro)
try:
    df_copy["Date"] = pd.to_datetime(df_copy["Date"], format="%Y/%m/%d")
except Exception as e:
    print("Erro ao converter 'Date':", e)

# 9. Substituindo '1900/01/01' por NaN e tentando novamente
df_copy["Date"] = df_copy["Date"].replace("1900/01/01", pd.NA)

# 10. Corrigindo valor '20201226' para '2020/12/26'
df_copy["Date"] = df_copy["Date"].replace("20201226", "2020/12/26")

# 11. Convertendo coluna 'Date' para datetime
df_copy["Date"] = pd.to_datetime(df_copy["Date"], errors="coerce")

# 12. Removendo registros com valores nulos (somente Date deve ter)
df_copy.dropna(inplace=True)

# 13. Resultado final
print("\nDataFrame final tratado:")
print(df_copy)
```
